# Univariate Linear Regression

Here I set `w` and `b` manually. Later, the algorithm will find the correct values on its own

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([2,10]) #features
y_train = np.array([15,65]) #targets
```

```python
i = 0
x_i = x_train[i]
y_i = y_train[i]
print(f"x: {x_i}, y: {y_i}")
```

```python
plt.scatter(x_train, y_train, marker='x', color='red')
plt.title('Univariate Linear Regression')
plt.xlabel('Features')
plt.ylabel('Targets')
plt.show()
```

```python
def compute_model(x, w, b): #linear function
    m = len(x)
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

```python
w = 6.25 #weight
b = 2.5 #bias
```

```python
tmp = compute_model(x_train, w, b)

plt.plot(x_train, tmp, color='blue', label='Prediction')
plt.scatter(x_train, y_train, marker='x', color='red', label='Training Data')
plt.title('Univariate Linear Regression')
plt.xlabel('Features')
plt.ylabel('Targets')
plt.legend()
plt.show()
```

```python
x_new = 6
y_hat = w * x_new + b
print(f"Prediction for x = {x_new}: {y_hat}")
```
